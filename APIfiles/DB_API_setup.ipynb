```python
# Uploading appointments data files to PostgreSQL database

import pandas as pd
from sqlalchemy import create_engine

# Step 1: Load CSV
encounters = pd.read_csv(r"C:\Users\aamreen_quantum-i\OneDrive\Desktop\Symptoms_checker\symptoms_checker\CSV_Data\encounters.csv")

# Step 2: Connection string (modify with your DB credentials)
USER = "mhealthadmin"       # e.g., postgres
PASSWORD = "mhealth*2025"   # e.g., 1234
HOST = "57.159.27.80"           # or your DB server IP
PORT = "5432"                # default postgres port
DBNAME = "mhealth_db"       # e.g., hospital_db

# Step 3: Create engine
engine = create_engine(f"postgresql+psycopg2://{USER}:{PASSWORD}@{HOST}:{PORT}/{DBNAME}")

# Step 4: Save DataFrame to Postgres (inside schema BHC)
encounters.to_sql(
    "encounters",  # table name in postgres
    engine,
    schema="BHC",           # schema name you created
    if_exists="replace",    # options: 'fail', 'replace', 'append'
    index=False             # don’t save pandas index
)

print("Data saved successfully to Postgres!")
```
